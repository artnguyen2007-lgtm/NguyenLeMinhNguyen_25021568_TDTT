```python
#Bai4
while True:
    try:
        h = int(input("Nhập độ cao h của tam giác (số nguyên dương): "))
        if h > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương.")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")

print(f"\nHình tam giác cân có độ cao h = {h}:")
for i in range(1, h + 1):
    spaces = " " * (h - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

```python
#Bai7
import math
while True:
    try:
        a = int(input("Nhập vào một số nguyên dương a: "))
        if a > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (lớn hơn 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
count = 0
limit = int(math.sqrt(a))
for i in range(1, limit + 1):
    if a % i == 0:
        if i * i == a:
            count += 1
        else:
            count += 2
print(f"\nSố ước dương của {a} là: {count}")
```

```python
#Bai8
import math
while True:
    try:
        a = int(input("Nhập số nguyên dương a: "))
        b = int(input("Nhập số nguyên dương b: "))
        if a > 0 and b > 0:
            break
        else:
            print("Vui lòng nhập cả hai số nguyên dương (lớn hơn 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
gcd_value = math.gcd(a, b)
print(f"\nƯớc chung lớn nhất (GCD) của {a} và {b} là: {gcd_value}")
print(f"Toàn bộ ước dương chung của {a} và {b} là:")
for i in range(1, gcd_value + 1):
    if gcd_value % i == 0:
        print(i)
```

```python
#Bai9
import math
while True:
    try:
        a = int(input("Nhập vào một số nguyên dương a: "))
        if a > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (lớn hơn 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
can_bac_hai = math.sqrt(a)
phan_nguyen = int(can_bac_hai)
if phan_nguyen == can_bac_hai:
    print(f"\n✅ KẾT QUẢ: {a} là SỐ CHÍNH PHƯƠNG.")
    print(f"Bởi vì {a} = {phan_nguyen} x {phan_nguyen}")
else:
    print(f"\n❌ KẾT QUẢ: {a} KHÔNG phải là số chính phương.")
```

```python
#Bai10
while True:
    try:
        a = int(input("Nhập vào một số nguyên dương a: "))
        if a > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (lớn hơn 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
if a == 1:
    print(f"\n❌ KẾT QUẢ: {a} KHÔNG phải là số hoàn hảo.")
else:
    sum_of_divisors = 0
    for i in range(1, a):
        if a % i == 0:
            sum_of_divisors += i
    if sum_of_divisors == a:
        print(f"\n✅ KẾT QUẢ: {a} là SỐ HOÀN HẢO.")
        print(f"Tổng các ước thực sự của nó là: {sum_of_divisors}")
    else:
        print(f"\n❌ KẾT QUẢ: {a} KHÔNG phải là số hoàn hảo.")
        print(f"Tổng các ước thực sự của nó là: {sum_of_divisors} (Khác {a})")
```

```python
#Bai12
while True:
    try:
        n = int(input("Nhập vào một số nguyên dương n: "))
        if n > 1:
            break
        else:
            print("Vui lòng nhập số nguyên dương n lớn hơn 1.")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
k = 0
current_sum = 0
previous_sum = 0
while True:
    k += 1
    previous_sum = current_sum
    current_sum = previous_sum + k
    if current_sum >= n:
        k_result = k - 1
        S_k_result = previous_sum
        break
print("\n--- KẾT QUẢ ---")
print(f"Giá trị n đã nhập: {n}")
print(f"Số nguyên dương k lớn nhất thỏa mãn S(k) < {n} là: k = {k_result}")
print(f"Với giá trị k này, tổng S({k_result}) = 1 + ... + {k_result} = {S_k_result}")
```

```python
#Bai13
while True:
    try:
        A = int(input("Nhập vào một số nguyên dương A: "))
        if A > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (A > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
n = 0
current_sum = 0.0
while current_sum <= A:
    n += 1
    current_sum += (1 / n)
print(f"Giá trị A đã nhập: {A}")
print(f"Số nguyên dương n nhỏ nhất thỏa mãn S_n > A là: n = {n}")
print(f"Với n = {n} thì S({n}) = {current_sum:.6f} > {A}")
```

```python
#Bai14
print("Bắt đầu nhập dãy số nguyên. Nhập -1 để kết thúc.")
while True:
    try:
        so_dau_tien = int(input("Nhập số thứ 1: "))
        if so_dau_tien != -1:
            so_lon_nhat = so_dau_tien
            so_nho_nhat = so_dau_tien
            break
        else:
            print("\n❌ Dãy số rỗng (chỉ nhập -1). Không thể tìm Min/Max.")
            exit() 
    except ValueError:
        print("❌ Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
so_luong_da_nhap = 1 

while True:
    so_luong_da_nhap += 1
    
    try:
        so_moi = int(input(f"Nhập số thứ {so_luong_da_nhap}: "))
    except ValueError:
        print("❌ Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
        so_luong_da_nhap -= 1 
        continue
    if so_moi == -1:
        print("\n--- KẾT THÚC NHẬP DỮ LIỆU ---")
        break    
    if so_moi > so_lon_nhat:
        so_lon_nhat = so_moi
    if so_moi < so_nho_nhat:
        so_nho_nhat = so_moi
print(f"Số lớn nhất trong dãy là: {so_lon_nhat}")
print(f"Số nhỏ nhất trong dãy là: {so_nho_nhat}")
```

```python
#Bai15
import math
def nhap_so_duong():
    while True:
        try:
            n = int(input("Nhập vào một số nguyên dương n: "))
            if n > 0:
                return n
            else:
                print("Vui lòng nhập số nguyên dương (n > 0).")
        except ValueError:
            print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
def dem_chu_so_chia_nguyen(n):
    if n == 0:
        return 1
    count = 0
    temp_n = n
    while temp_n > 0:
        temp_n //= 10
        count += 1
    return count
n = nhap_so_duong()
so_chu_so_1 = dem_chu_so_chia_nguyen(n)
print(f"Số chữ số của {n} là: {so_chu_so_1}")
```

```python
#Bai16
while True:
    try:
        n = int(input("Nhập vào một số nguyên dương n: "))
        if n > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (n > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
count_chan = 0
count_le = 0
temp_n = n
while temp_n > 0:
    chu_so = temp_n % 10
    if chu_so % 2 == 0:
        count_chan += 1
    else:
        count_le += 1
    temp_n //= 10
print(f"Số đã nhập: {n}")
print(f"Số lượng chữ số chẵn: {count_chan}")
print(f"Số lượng chữ số lẻ: {count_le}")
```

```python
#Bai18
while True:
    try:
        n = int(input("Nhập vào một số nguyên dương n: "))
        if n > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (n > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
def is_power_of_three_loop(num):
    if num == 1:
        return True
    if num <= 0 or num % 3 != 0:
        return False
    while num > 1:
        if num % 3 != 0:
            return False
        num //= 3
    return num == 1
if is_power_of_three_loop(n):
    print(f"\n✅ KẾT QUẢ: {n} là số có dạng 3^k.")
else:
    print(f"\n❌ KẾT QUẢ: {n} KHÔNG phải là số có dạng 3^k.")
```

```python
#Bai19
while True:
    try:
        A = int(input("Nhập vào một số nguyên dương A: "))
        if A > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (A > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
f1 = 1
f2 = 1
fib_max_found = 1
if A == 1:
    fib_max_found = 1
elif A > 1:
    while True:
        f_next = f1 + f2
        if f_next > A:
            break
        fib_max_found = f_next
        f1 = f2
        f2 = f_next
print(f"Giá trị A đã nhập: {A}")
print(f"Số Fibonacci lớn nhất nhưng không vượt quá A là: {fib_max_found}")
```

```python
#Bai20
while True:
    try:
        n = int(input("Nhập vào một số nguyên dương n: "))
        if n > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (n > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
count_nguyen_to = 0
temp_n = n
NGUYEN_TO_SET = {2, 3, 5, 7}
while temp_n > 0:
    chu_so = temp_n % 10
    if chu_so in NGUYEN_TO_SET:
        count_nguyen_to += 1
    temp_n //= 10
print(f"Số đã nhập: {n}")
print(f"Các chữ số nguyên tố là: 2, 3, 5, 7")
print(f"Số lượng chữ số nguyên tố trong {n} là: {count_nguyen_to}")
```
