```python
gio_hang = [
    {"ten": "Ao thun", "so_luong": 2, "don_gia": 150000},
    {"ten": "quan dai", "so_luong":1, "don_gia":360000},
]

tong = 0

for mat_hang in gio_hang:
    so_tien = mat_hang["so_luong"] * mat_hang["don_gia"]
    tong += so_tien

print ("Tong so tien phai tra", tong,  "VND")   
```
