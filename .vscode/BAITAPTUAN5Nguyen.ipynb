```python
#W5A1
def Max_Of_Two(a: int, b: int) -> int:
    if a > b:
        return a
    else:
        return b
try:
    a_input = input("Nhập số nguyên A: ")
    a = int(a_input)
    b_input = input("Nhập số nguyên B: ")
    b = int(b_input)
    so_lon_nhat = Max_Of_Two(a, b)
    print(f"Số lớn hơn giữa {a} và {b} là: {so_lon_nhat}")
except ValueError:
    print("Lỗi: Vui lòng chỉ nhập số nguyên.")      
```

```python
#W5A2
from typing import Tuple
def swap(a: int, b: int) -> Tuple[int, int]:
    a, b = b, a 
    return (a, b)
try:
    a_input = input("Nhập số nguyên A: ")
    a = int(a_input)
    b_input = input("Nhập số nguyên B: ")
    b = int(b_input)
    print(f"\n[TRƯỚC KHI HOÁN ĐỔI] A = {a}, B = {b}") 
    a, b = swap(a, b)
    print(f"[SAU KHI HOÁN ĐỔI] A = {a}, B = {b}") 
except ValueError:
    print("Lỗi: Đầu vào không hợp lệ. Vui lòng chỉ nhập số nguyên.")
```

```python
#W5A3
import math
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    limit = int(math.sqrt(n)) + 1   
    for i in range(2, limit):
        if n % i == 0:
            return False
    return True
try:
    n_input = input("Nhập một số nguyên dương để kiểm tra: ")
    number = int(n_input)
    if is_prime(number):
        print(f"\n✅ Kết quả: Số {number} là SỐ NGUYÊN TỐ.")
    else:
        print(f"\n❌ Kết quả: Số {number} KHÔNG phải là số nguyên tố.")
except ValueError:
    print("Lỗi: Đầu vào không hợp lệ. Vui lòng chỉ nhập số nguyên.")   
```

```python
#W5A4
def is_perfect_number(n: int) -> bool:
    if n <= 1:
        return False        
    sum_of_divisors = 0
    for i in range(1, (n // 2) + 1):
        if n % i == 0:
            sum_of_divisors += i
    return sum_of_divisors == n
try:
    n_input = input("Nhập một số nguyên dương để kiểm tra số hoàn hảo: ")
    number = int(n_input)
    if is_perfect_number(number):
        print(f"\n✅ Kết quả: Số {number} là SỐ HOÀN HẢO.")
    else:
        print(f"\n❌ Kết quả: Số {number} KHÔNG phải là số hoàn hảo.")
except ValueError:
    print("Lỗi: Đầu vào không hợp lệ. Vui lòng chỉ nhập số nguyên.")   
```

```python
#W5A5
from typing import List
def find_first_occurrence(data_list: List[int], k: int) -> int:    
    for index in range(len(data_list)):
        if data_list[index] == k:
            return index
    return -1
def get_list_from_input() -> List[int]:
    while True:
        try:
            input_str = input("Nhập danh sách các số nguyên (cách nhau bằng dấu phẩy hoặc khoảng trắng): ")
            elements = [int(x.strip()) for x in input_str.replace(',', ' ').split() if x.strip()]
            return elements
        except ValueError:
            print("Lỗi: Vui lòng đảm bảo bạn chỉ nhập số nguyên và sử dụng dấu phẩy hoặc khoảng trắng để phân cách.")
        except Exception as e:
            print(f"Lỗi không xác định: {e}")
my_list = get_list_from_input()
print(f"Danh sách đã nhập: {my_list}")
while True:
    try:
        k_input = input("Nhập số nguyên dương K cần tìm: ")
        k = int(k_input)
        if k < 0:
            print("Vui lòng nhập số nguyên dương (hoặc 0 nếu muốn tìm 0).")
            continue
        break
    except ValueError:
        print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")
position = find_first_occurrence(my_list, k)
if position != -1:
    print(f"Giá trị k={k} được tìm thấy lần đầu tiên tại vị trí (chỉ mục) {position}.")
else:
    print(f"Không tìm thấy giá trị k={k} trong danh sách. Hàm trả về: {position}")
```

```python
#W5A6
def calculate_factorial(n: int) -> int:
    if n < 0:
        raise ValueError("Giai thừa chỉ được tính cho số nguyên không âm (n >= 0).")
    if n == 0 or n == 1:
        return 1
    factorial = 1
    for i in range(2, n + 1):
        factorial *= i      
    return factorial
try:
    n_input = input("Nhập một số nguyên không âm n để tính giai thừa (n!): ")
    n = int(n_input)
    result = calculate_factorial(n)   
    print(f"\n✅ Kết quả: Giai thừa của {n} ({n}!) là: {result}")
except ValueError as e:
    print(f"❌ Lỗi: {e}")
except Exception:
    print("❌ Lỗi: Vui lòng đảm bảo bạn nhập một số nguyên hợp lệ.")
```

```python
#W5A7
from typing import Optional
def calculator(num1: float, num2: float, operator: str) -> Optional[float]:
    if operator == '+':
        result = num1 + num2
    elif operator == '-':
        result = num1 - num2
    elif operator == '*':
        result = num1 * num2
    elif operator == '/':
        if num2 == 0:
            return "Lỗi: Chia cho 0"
        result = num1 / num2
    else:
        return "Lỗi: Toán tử không hợp lệ"
    return round(result, 2)
def run_calculator_interactive():
    print("--- MÁY TÍNH BỎ TÚI ĐƠN GIẢN ---")
    while True:
        try:
            num1_input = input("Nhập số thứ nhất (num1): ")
            num1 = float(num1_input)
            break
        except ValueError:
            print("Lỗi: Vui lòng nhập một số hợp lệ.")
    while True:
        try:
            num2_input = input("Nhập số thứ hai (num2): ")
            num2 = float(num2_input)
            break
        except ValueError:
            print("Lỗi: Vui lòng nhập một số hợp lệ.")
    valid_operators = ['+', '-', '*', '/']
    while True:
        operator = input("Nhập toán tử (+, -, *, /): ")
        if operator in valid_operators:
            break
        else:
            print(f"Lỗi: Toán tử '{operator}' không được hỗ trợ. Vui lòng chọn một trong: {', '.join(valid_operators)}.")
    calculation_result = calculator(num1, num2, operator)
    if isinstance(calculation_result, str) and calculation_result.startswith("Lỗi:"):
        print(f"{calculation_result}")
    elif calculation_result is not None:
        print(f"{num1} {operator} {num2} = {calculation_result}")
run_calculator_interactive()   
```

```python
#W5A8
def hamming_distance(x: int, y: int) -> int:
    xor_result = x ^ y
    distance = bin(xor_result).count('1')  
    return distance
def run_hamming_interactive():
    print("--- TÍNH KHOẢNG CÁCH HAMMING ---")
    while True:
        try:
            x_input = input("Nhập số nguyên X: ")
            x = int(x_input)
            break
        except ValueError:
            print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")
    while True:
        try:
            y_input = input("Nhập số nguyên Y: ")
            y = int(y_input)
            break
        except ValueError:
            print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")
    distance = hamming_distance(x, y)
    print(f"Số X: {x} (Binary: {x:b})")
    print(f"Số Y: {y} (Binary: {y:b})")
    print(f"Khoảng cách Hamming giữa {x} và {y} là: {distance}")
run_hamming_interactive()   
```

```python
#W5A9
def sum_digits(n: int) -> int:   
    if n < 0:
        n = abs(n) 
    sum_val = 0
    current_number = n
    while current_number > 0:
        digit = current_number % 10
        sum_val += digit
        current_number //= 10     
    return sum_val
def run_sum_digits_interactive():
    print("--- TÍNH TỔNG CÁC CHỮ SỐ ---")
    while True:
        try:
            n_input = input("Nhập một số nguyên dương n: ")
            n = int(n_input)
            if n < 0:
                print(f"Lưu ý: Bạn đã nhập số âm, sẽ tính tổng chữ số của giá trị tuyệt đối |{n}|.")        
            break
        except ValueError:
            print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")
    total_sum = sum_digits(n)
    print(f"Tổng các chữ số của số {n} là: {total_sum}")
run_sum_digits_interactive()   
```

```python
#W5A10
from typing import Dict, Optional
def are_isomorphic(a: str, b: str) -> bool:
    if len(a) != len(b):
        return False
    map_a_to_b: Dict[str, str] = {}
    map_b_to_a: Dict[str, str] = {}   
    n = len(a)
    for i in range(n):
        char_a = a[i]
        char_b = b[i]
        if char_a in map_a_to_b:
            if map_a_to_b[char_a] != char_b:
                return False
        elif char_b in map_b_to_a:
            return False 
        else:
            map_a_to_b[char_a] = char_b
            map_b_to_a[char_b] = char_a           
    return True
def run_isomorphic_interactive():
    print("--- KIỂM TRA TÍNH ĐẲNG CẤU GIỮA HAI TỪ ---")   
    word_a: Optional[str] = None
    word_b: Optional[str] = None
    word_a = input("Nhập Từ A: ")
    while True:
        word_b = input("Nhập Từ B: ")
        if len(word_a) != len(word_b):
            print(f"Lỗi: Hai từ phải có cùng độ dài ({len(word_a)}). Vui lòng nhập lại Từ B.")
        else:
            break
    isomorphic = are_isomorphic(word_a, word_b)
    if isomorphic:
        print(f"✅ Từ '{word_a}' và '{word_b}' LÀ đẳng cấu.")
    else:
        print(f"❌ Từ '{word_a}' và '{word_b}' KHÔNG đẳng cấu.")
run_isomorphic_interactive()    
```
