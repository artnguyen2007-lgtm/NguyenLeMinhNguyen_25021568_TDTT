```python
#W3A4
a = int(input("nhap so a: "))
b = int(input("nhap so b: "))
ket_qua = a // b
print("Ket qua chia lam tron xuong:", ket_qua)

x = a//b
y = round(a/b, 1)
if y - x < 0.5:
    print("Ket qua chia lam tron xuong:", x)
```

```python
#W3A5
a = int(input("nhap so a: "))
b = int(input("nhap so b: "))
ket_qua = (a + b - 1) // b
print("Ket qua chia lam tron len:", ket_qua)

x = a//b
y = round(a/b, 1)
if y - x >= 0.5:
    print("Ket qua chia lam tron len:", x + 1)
```

```python
#W3A11
a, b, c = map(int, input("nhap 3 so nguyen:").split())
if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
    print("3 so tren tao thanh canh cua tam giac vuong")
elif a == b or a == c or b == c:
    print("3 so tren tao thanh canh cua tam giac can")
elif a == b == c:
    print("3 so tren tao thanh canh cua tam giac deu")
elif a+b > c and a+c > b and b+c > a:
    print("3 so tren tao thanh canh cua tam giac thuong")
else:
    print("3 so tren khong tao thanh canh cua tam giac")
```

```python
#W3A16
import math
a = float(input("nhap so thuc a: "))
if a > 0:
    tron_len = int(a) if a == int(a) else int(a) + 1
else:
    tron_len = int(a)

tron_xuong = int(a) if a >= 0 else (int(a) - 1 if a != int(a) else int(a))

phan_thap_phan = a - int(a)
if phan_thap_phan >= 0.5:
    tron_gan_nhat = int(a) + 1
elif phan_thap_phan <= -0.5:
    tron_gan_nhat = int(a) - 1
else:
    tron_gan_nhat = int(a)

print(tron_len, tron_xuong, tron_gan_nhat)
```
