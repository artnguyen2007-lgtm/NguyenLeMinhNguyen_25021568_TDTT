```python
#W4A2
import math

def kiem_tra_so_nguyen_to(n):
    """Kiểm tra xem một số nguyên dương có phải là số nguyên tố hay không."""
    
    if n <= 1:
        return False
        
    if n == 2:
        return True
        
    if n % 2 == 0:
        return False
        
    limit = int(math.sqrt(n))
    
    for i in range(3, limit + 1, 2):
        if n % i == 0:
            return False
            
    return True
while True:
    try:
        n = int(input("Nhập vào một số nguyên dương: "))
        
        if n > 0:
            break
        else:
            print(f"❌ Số {n} không phải số nguyên dương. Vui lòng nhập lại.")
            
    except ValueError:
        print("❌ Đầu vào không hợp lệ. Vui lòng chỉ nhập số nguyên.")

if kiem_tra_so_nguyen_to(n):
    print(f"✅ Số {n} là SỐ NGUYÊN TỐ.")
else:
    print(f"❌ Số {n} KHÔNG phải là số nguyên tố.")
```

```python
#W4A6
import math
def kiem_tra_so_nguyen_to(n):
    """Trả về True nếu n là số nguyên tố, ngược lại trả về False."""
    
    if n <= 1:
        return False
        
    if n == 2:
        return True
        
    if n % 2 == 0:
        return False
        
    limit = int(math.sqrt(n))
    for i in range(3, limit + 1, 2):
        if n % i == 0:
            return False
            
    return True

def tinh_tong_so_nguyen_to_trong_doan(a, b):
    tong_nguyen_to = 0
    
    for num in range(a, b + 1):
        if kiem_tra_so_nguyen_to(num):
            tong_nguyen_to += num
            
    return tong_nguyen_to

def main():
    print("--- CHƯƠNG TRÌNH TÍNH TỔNG SỐ NGUYÊN TỐ ---")
    while True:
        try:
            a = int(input("Nhập số nguyên dương a: "))
            b = int(input("Nhập số nguyên dương b (b >= a): "))
            
            if a > 0 and b > 0 and a <= b:
                break
            else:
                print("Lỗi: Vui lòng nhập a, b là số nguyên dương và a <= b.")
        except ValueError:
            print("Lỗi: Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")

    tong = tinh_tong_so_nguyen_to_trong_doan(a, b)
    
    print(f"Các số nguyên tố trong đoạn [{a}, {b}] có tổng là: {tong}")

if __name__ == "__main__":
    main()
```

```python
#W4A7
import math

def tim_uoc_nguyen_to_lon_nhat(n):
    """Tìm ước số nguyên tố lớn nhất của số nguyên dương n."""
    
    max_prime_divisor = 1
    
    while n % 2 == 0:
        max_prime_divisor = 2
        n //= 2
        
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        while n % i == 0:
            max_prime_divisor = i
            n //= i
            
    if n > 1:
        max_prime_divisor = n
        
    return max_prime_divisor

while True:
    try:
        n = int(input("Nhập vào số nguyên dương n (n >= 2): "))
        if n >= 2:
            break
        else:
            print("Vui lòng nhập số nguyên dương n lớn hơn hoặc bằng 2.")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
uoc_lon_nhat = tim_uoc_nguyen_to_lon_nhat(n)

print(f"Số đã nhập: {n}")
print(f"Ước số nguyên tố lớn nhất của {n} là: {uoc_lon_nhat}")
```

```python
#W4A8
def is_palindrome(n):
    """Kiểm tra xem một số nguyên có phải là số đối xứng (palindrome) hay không."""
    s = str(n)
    return s == s[::-1]
def reverse_number(n):
    """Đảo ngược các chữ số của một số nguyên n."""
    s = str(n)
    s_reversed = s[::-1]
    return int(s_reversed)
while True:
    try:
        n = int(input("Nhập vào một số nguyên dương n: "))
        if n > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (n > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
so_buoc = 0
current_n = n
print("\n--- QUÁ TRÌNH LẶP ---")
while not is_palindrome(current_n):
    reversed_n = reverse_number(current_n)
    next_n = current_n + reversed_n
    so_buoc += 1
    print(f"Bước {so_buoc}: {current_n} + {reversed_n} = {next_n}")
    current_n = next_n
print(f"Số bước lặp: {so_buoc}")
print(f"Giá trị palindrome thu được: {current_n}")
```

```python
#W4A9
import math
def kiem_tra_chu_so_khac_nhau(so):
    """Kiểm tra xem tất cả các chữ số của 'so' có đôi một khác nhau hay không."""
    s = str(so)
    chu_so_da_xet = set()   
    for chu_so in s:
        if chu_so in chu_so_da_xet:
            return False
        chu_so_da_xet.add(chu_so)
        
    return True
while True:
    try:
        n = int(input("Nhập vào số nguyên dương n: "))
        if n > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (n > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
danh_sach_ket_qua = []
can_bac_hai_max = int(math.sqrt(n))
for k in range(1, can_bac_hai_max + 1):
    so_chinh_phuong = k * k
    if kiem_tra_chu_so_khac_nhau(so_chinh_phuong):
        danh_sach_ket_qua.append(so_chinh_phuong)
if danh_sach_ket_qua:
    print(*danh_sach_ket_qua, sep=' ')
else:
    print("(no number)")
```

```python
#W4A10          
def tinh_do_dai_collatz(n):
    """Tính độ dài dãy Collatz cho một số nguyên dương n."""
    if n <= 0:
        return 0
    length = 1
    current = n
    
    while current > 1:
        if current % 2 == 0:
            current //= 2
        else:
            current = 3 * current + 1
        length += 1
        
    return length
while True:
    try:
        n = int(input("Nhập vào số nguyên dương n: "))
        if n > 0:
            break
        else:
            print("Vui lòng nhập số nguyên dương (n > 0).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
max_length = 0
so_co_max_length = 0
for x in range(1, n + 1):
    do_dai_hien_tai = tinh_do_dai_collatz(x)
    if do_dai_hien_tai > max_length:
        max_length = do_dai_hien_tai
        so_co_max_length = x
print(f"Số n đã nhập: {n}")
print(f"Số x trong [1, {n}] có độ dài dãy Collatz lớn nhất là:")
print(f"{so_co_max_length} {max_length}")
```

```python
#W4A11
import math
def dem_so_luong_uoc(m):
    """Đếm số lượng ước của một số nguyên dương m (ước lẻ)."""
    count = 0
    limit = int(math.sqrt(m))   
    for i in range(1, limit + 1):
        if m % i == 0:
            if i * i == m:
                count += 1
            else:
                count += 2
    return count
def dem_uoc_so_chan(n):
    if n == 1:
        return 0
    k = 0 
    m = n 
    while m % 2 == 0:
        k += 1
        m //= 2
    so_luong_uoc_le = dem_so_luong_uoc(m)
    so_luong_uoc_chan = k * so_luong_uoc_le
    
    return so_luong_uoc_chan
MAX_N = 1000000
while True:
    try:
        n = int(input(f"Nhập vào số nguyên dương n (n < {MAX_N}): "))
        if 1 <= n < MAX_N:
            break
        else:
            print(f"Vui lòng nhập n trong khoảng [1, {MAX_N - 1}].")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
ket_qua = dem_uoc_so_chan(n)
print(f"Số đã nhập: {n}")
print(f"Số lượng ước số chẵn của {n} là: {ket_qua}")
```

```python
#W4A12
LAI_SUAT_HANG_THANG = 0.007
while True:
    try:
        X = float(input("Nhập số tiền gốc gửi ban đầu (X đồng): "))
        if X <= 0:
            print("Số tiền gốc phải là số dương.")
            continue
            
        N = int(input("Nhập số tháng gửi (N tháng): "))
        if N < 0:
            print("Số tháng gửi không thể là số âm.")
            continue
            
        break
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số.")
he_so_lai = (1 + LAI_SUAT_HANG_THANG) ** N
so_tien_cuoi_cung = X * he_so_lai
so_tien_rut_duoc = int(so_tien_cuoi_cung)
print(f"Số tiền gốc ban đầu (X): {X:,.0f} đồng")
print(f"Lãi suất: {LAI_SUAT_HANG_THANG * 100:.1f}% / tháng")
print(f"Thời gian gửi: {N} tháng")
print(f"Số tiền cả gốc và lãi sau {N} tháng (làm tròn xuống): {so_tien_rut_duoc:,.0f} đồng")
```

```python
#W4A13
def tinh_tong_uoc_thuc_su(n):
    """Tính tổng các ước dương của n, trừ chính n."""
    if n <= 1:
        return 0     
    tong = 1 
    limit = int(n**0.5) 
    for i in range(2, limit + 1):
        if n % i == 0:
            tong += i
            uoc_doi = n // i
            if uoc_doi != i:
                tong += uoc_doi     
    return tong
while True:
    try:
        a = int(input("Nhập số nguyên dương a: "))
        b = int(input("Nhập số nguyên dương b: "))
        if a > 0 and b > 0 and a != b:
            break
        else:
            print("Vui lòng nhập hai số nguyên dương khác nhau (a, b > 0, a != b).")
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
tong_uoc_a = tinh_tong_uoc_thuc_su(a)
tong_uoc_b = tinh_tong_uoc_thuc_su(b)
if tong_uoc_a == b and tong_uoc_b == a:
    print("true")
else:
    print("false")
```

```python
#W4A14
def tim_uoc_chung_lon_nhat(m, n):
    """Sử dụng thuật toán Euclidean để tìm GCD của hai số."""
    while n != 0:
        r = m % n
        m = n
        n = r
    return m
while True:
    try:
        m = int(input("Nhập số nguyên dương m: "))
        n = int(input("Nhập số nguyên dương n: "))
        if m > 0 and n > 0:
            break
        else:
            print("Vui lòng nhập cả hai số nguyên dương (m, n > 0).")       
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
gcd_result = tim_uoc_chung_lon_nhat(m, n)
print(f"Ước chung lớn nhất của {m} và {n} là: {gcd_result}")
```

```python
#W4A15
def giai_bai_toan_ga_cho(tong_so_con, tong_so_chan):
    if tong_so_con <= 0 or tong_so_chan <= 0:
        return "invalid"
    if tong_so_chan % 2 != 0:
        return "invalid"
    min_chan = 2 * tong_so_con
    max_chan = 4 * tong_so_con
    if not (min_chan <= tong_so_chan <= max_chan):
        return "invalid"
    tu_so = tong_so_chan - (2 * tong_so_con)
    so_cho = tu_so // 2  
    so_ga = tong_so_con - so_cho
    if so_ga >= 0 and so_cho >= 0:
        return f"Số gà: {so_ga}, Số chó: {so_cho}"
    else:
        return "invalid"
while True:
    try:
        con = int(input("Nhập TỔNG SỐ CON: "))
        chan = int(input("Nhập TỔNG SỐ CHÂN: "))
        break
    except ValueError:
        print("Đầu vào không hợp lệ. Vui lòng nhập số nguyên.")
ket_qua = giai_bai_toan_ga_cho(con, chan)
print(ket_qua)
```
