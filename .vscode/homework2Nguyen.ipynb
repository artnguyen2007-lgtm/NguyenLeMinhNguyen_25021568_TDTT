```python
#Bai1
a = float(input("Nhap chieu dai cua hinh chu nhat: "))
b = float(input("Nhap chieu rong cua hinh chu nhat: "))
if a <= 0 or b <= 0:
    print("Chieu dai va chieu rong phai la so duong!")
else:
    chuvi = 2 * (a + b)
    dientich = a * b
    print("Chu vi hinh chu nhat la:", chuvi)
    print("Dien tich hinh chu nhat la:", dientich)
```

```python
#Bai2
r = float(input("Nhap ban kinh hinh tron: "))
if r <= 0:
    print("Ban kinh phai la so duong!")
else:
    import math
    chuvi = 2 * math.pi * r
    dientich = math.pi * r**2
    print("Chu vi hinh tron la:", chuvi)
    print("Dien tich hinh tron la:", dientich)   
```

```python
#bai3
import math
a = int(input("nhap so a:"))
b = int(input("nhap so b:"))
c = int(input("nhap so c:"))
p = (a + b + c) / 2
dientich = math.sqrt(p * (p - a) * (p - b) * (p - c))
dientich = round(dientich, 2)
if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
    print("tao thanh canh cua tam giac vuong")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
elif a == b or a == c or b == c:
    print("tao thanh canh cua tam giac can")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
elif a == b == c:
    print("tao thanh canh cua tam giac deu")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
elif a+b > c and a+c > b and b+c > a:
    print("tao thanh canh cua tam giac thuong")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
else:
    print("3 so tren khong tao thanh canh cua tam giac")
```

```python
#Bai4
def giai_phuong_trinh_bac_nhat(a, b):
    print(f"\nGiải phương trình: {a}x + {b} = 0")
    
    if a == 0:
        if b == 0:
            print("Kết quả: Phương trình có **vô số nghiệm** (tập hợp số thực R).")
        else:
            print("Kết quả: Phương trình **vô nghiệm**.")
    
    else:
        x = -b / a
        print(f"Kết quả: Phương trình có **nghiệm duy nhất**:")
        print(f"x = -({b}) / {a} ≈ {x:.4f}") 

print("--- Giải Phương Trình Bậc Nhất (ax + b = 0) ---")

try:
    print("\n--- Nhập Giá Trị Hệ Số ---")
    a_user = float(input("Nhập hệ số a: "))
    b_user = float(input("Nhập hệ số b: "))
    
    giai_phuong_trinh_bac_nhat(a_user, b_user)

except ValueError:
    print("\nLỗi: Vui lòng nhập các giá trị số hợp lệ cho a và b.")
```

```python
#Bai5
import math

def giai_phuong_trinh_bac_hai(a, b, c):
    print(f"\nGiải phương trình: {a}x^2 + {b}x + {c} = 0")
    
    if a == 0:
        if b == 0:
            if c == 0:
                print("Kết quả: Phương trình trở thành 0 = 0. Có **vô số nghiệm** (tập hợp số thực R).")
            else:
                print("Kết quả: Phương trình trở thành c = 0 (với c != 0). **Vô nghiệm**.")
        else:
            x = -c / b
            print(f"Kết quả: Phương trình trở thành bậc nhất. Có **nghiệm duy nhất**: x = {-c}/{b} ≈ {x:.4f}")
        return

    delta = b**2 - 4*a*c
    print(f"Biệt thức Delta (Δ) = {b}^2 - 4 * {a} * {c} = {delta}")
    
    if delta < 0:
        print("Kết quả: Δ < 0. Phương trình **vô nghiệm** trong tập số thực.")
    elif delta == 0:
        x = -b / (2 * a)
        print("Kết quả: Δ = 0. Phương trình có **nghiệm kép**:")
        print(f"x1 = x2 = -({b}) / (2 * {a}) ≈ {x:.4f}")
        
    else:
        sqrt_delta = math.sqrt(delta)
        x1 = (-b + sqrt_delta) / (2 * a)
        x2 = (-b - sqrt_delta) / (2 * a)
        
        print("Kết quả: Δ > 0. Phương trình có **hai nghiệm phân biệt**:")
        print(f"x1 = (-({b}) + sqrt({delta})) / (2 * {a}) ≈ {x1:.4f}")
        print(f"x2 = (-({b}) - sqrt({delta})) / (2 * {a}) ≈ {x2:.4f}")


print("--- Giải Phương Trình Bậc Hai (ax^2 + bx + c = 0) ---")

try:
    print("\n--- Nhập Giá Trị Hệ Số ---")
    a_user = float(input("Nhập hệ số a: "))
    b_user = float(input("Nhập hệ số b: "))
    c_user = float(input("Nhập hệ số c: "))
    
    giai_phuong_trinh_bac_hai(a_user, b_user, c_user)

except ValueError:
    print("\nLỗi: Vui lòng nhập các giá trị số hợp lệ cho a, b và c.")
```

```python
#Bai8
a, b, c = map(float, input("nhap 3 so nguyen:").split())
d, e, f = map(float, input("nhap 3 so nguyen:").split())
Det = a * e - b * d
if Det == 0:
    print("He phuong trinh vo nghiem hoac vo so nghiem")
else:
    Detx = c * e - b * f
    Dety = a * f - c * d
    x = Detx / Det
    y = Dety / Det
    print("Nghiem cua he phuong trinh la: x = ",x, " y = ", y)
```

```python
#Bai10
Ix, Iy = map(int, input("nhap tam cua duong tron:").split())
R = int(input("nhap ban kinh duong tron:"))
Ax, Ay = map(int, input("nhap toa do diem A:").split())
if (Ax - Ix)**2 + (Ay - Iy)**2 == R**2:
    print("Diem A nam tren duong tron")
else:
    print("Diem A khong nam tren duong tron")
```
